```python
import sys
```

```python
def read_input():
    data = sys.stdin.readline
    n, m = map(int, data().split())
    a = list(map(int, data().split()))
    c = list(map(int, data().split()))
    return n, m, a, c
```

```python
### 1. USING ORTOOL
from ortools.sat.python import cp_model

def solve_ortools(n, m, a, c):
    model = cp_model.CpModel()
    x = []
    for i in range(n):
        x.append([model.new_bool_var(f"x[{i}][{j}]") for j in range(m)])
    for j in range(m):
        model.add(sum(x[i][j] * a[i] for i in range(n)) <= c[j])
    for i in range(n):
        model.add(sum(x[i][j] for j in range(m)) == 1)

    solver = cp_model.CpSolver()
    status = solver.solve(model)
    solver.parameters.max_time_in_seconds = 5

    arrange = [[] for _ in range(m)]
    for j in range(m):
        for i in range(n):
            if solver.value(x[i][j]) == 1:
                arrange[j].append(i)
    for index, items in enumerate(arrange):
        print(f"Các đồ vật có trong túi {index} là:", *items)
```

```python
### 2. USING GREEDY ALGORITHM
import heapq
def greedy(n, m, a, c):
    order = [-1] * n
    items = [(weight, i) for i, weight in enumerate(a)]
    items.sort(key=lambda x: x[0], reverse=True)
    heap = []
    for i, contain in enumerate(c):
        heapq.heappush(heap, (-contain, i))

    for weight, item in items:
        (contain_neg, bag_idx) = heapq.heappop(heap)
        contain = -contain_neg
        if contain >= weight:
            order[item] = bag_idx
            contain -= weight
            heapq.heappush(heap, (-contain, bag_idx))
        else:
            heapq.heappush(heap, (contain_neg, bag_idx))

    return order
```

```python
def main():
    f = open("input.txt", "r")
    sys.stdin = f
    n, m, a, c = read_input()
    solve_ortools(n, m, a, c)
    f.close()
```

```python
main()
```
