```python
import sys
from ortools.sat.python import cp_model
```

```python
def read_input():
    data = list(map(int, sys.stdin.read().split()))
    N, W, H = data[0], data[1], data[2]
    idx = 3
    w, h = [], []
    for _ in range(n):
        w.append(data[idx]); idx += 1
        h.append(data[idx]); idx += 1
    return N, W, H, w, h
```

```python
def solve_ortools(N, W, H, w, h):
    model = cp_model.CpModel()
    solver = cp_model.CpSolver()

    x_vars, y_vars, rotations = [], [], []
    x_intervals, y_intervals = [], []
    for i in range(N):
        r_i = model.new_bool_var(f"r_{i}")
        effective_h_i = model.new_int_var(min(h[i], w[i]), max(h[i], w[i]), f"effective_h_{i}")
        effective_w_i = model.new_int_var(min(h[i], w[i]), max(h[i], w[i]), f"effective_w_{i}")

        model.add(effective_h_i == (1 - r_i) * h_i + r_i * w_i)
        model.add(effective_w_i == (1 - r_i) * w_i + r_i * h_i)

        x_i = model.new_int_var(0, W, f"x_{i}")
        y_i = model.new_int_var(0, H, f"x_{i}")
        model.add(x_i + effective_w_i <= W)
        model.add(y_i + effective_h_i <= H)

        x_vars.append(x_i)
        y_vars.append(y_i)
        rotations.append(r_i)

        x_interval = model.new_interval_var(x_i, effective_w_i, x_i + effective_w_i, f"x_interval_{i}")
        y_interval = model.new_interval_var(y_i, effective_h_i, y_i + effective_h_i, f"y_interval_{i}")

        x_intervals.append(x_interval)
        y_intervals.append(y_interval)

    model.AddNoOverlap2D(x_intervals, y_intervals)
    solver.parameters.max_time_in_seconds = 3
    status = solver.solve(model)

    if status in (cp_model.OPTIMAL, cp_model.FEASIBLE):
        print("Tìm thấy phương án xếp hàng thành công")
    else:
        print("Không có phương án nào thỏa mãn")
```

```python
def main():
    f = open("input.txt", "r")
    sys.stdin = f
    N, W, H, w, h = read_input()
    solve_ortools(N, W, H, w, h)
    f.close()
```

```python
main()
```
