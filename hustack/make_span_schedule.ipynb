```python
import sys
from collections import deque

def read_input():
    input = sys.stdin.readline
    n, m = map(int, input().split())
    d = list(map(int, input().split()))
    Q = []
    adj = [[] for _ in range(n)]
    num_before = [0] * n
    for _ in range(m):
        a, b = map(int, input().split())
        Q.append((a - 1, b - 1))
        num_before[b - 1] += 1
        adj[a - 1].append(b - 1)
    return n, m, d, Q, num_before, adj

def solve(n, m, d, Q, num_before, adj):
    dq = deque()
    
    for i, num in enumerate(num_before):
        if num == 0:
            dq.append(i)

    lst = []
    while dq:
        i = dq.popleft()
        lst.append(i)
        for u in adj[i]:
            num_before[u] -= 1
            if num_before[u] == 0:
                dq.append(u)

    deadline = [0] * n
    res = 0
    for idx in lst:
        if deadline[idx] + d[idx] > res:
            res = deadline[idx] + d[idx]
        for u in adj[idx]:
            deadline[u] = max(deadline[u], deadline[idx] + d[idx])

    return res

def main():
    f = open("input.txt", "r")
    sys.stdin = f
    n, m, d, Q, num_before, adj = read_input()
    res = solve(n, m, d, Q, num_before, adj)
    print(res)
    f.close()
```

```python
main()
```
